````python
# Lets start with some imports
import pandas as pd
import numpy as np
import re
import json
import sys
from pathlib import Path
from textwrap import dedent

proj_root = Path('/Users/liekevanson/Documents/Projects/post_mt_review').resolve()
if str(proj_root) not in sys.path:
    sys.path.insert(0, str(proj_root))

# -------------------------------------------------
# Configuration
# -------------------------------------------------
from paths import RESULT_TABLES, RAW_JSON_DIR, DATA_DIR
````

````python
table_file = Path(DATA_DIR) / 'latex_input_data' / 'vanRoestel18.tex'

with open(table_file, 'r') as f:
    latex = f.read()

# Keep only table rows (skip headers, hlines, captions)
raw_lines = latex.splitlines()
lines = [
    ln for ln in raw_lines
    if '&' in ln
    and r'\hline' not in ln
    and 'Name &' not in ln
    and 'tabular' not in ln
    and 'caption' not in ln
    and 'contcaption' not in ln
    and 'noalign' not in ln
]


def parse_number(chunk: str) -> float:
    """Extract the first numeric token from a LaTeX cell."""
    match = re.search(r'[-+]?\d+(?:\.\d+)?', chunk)
    if not match:
        raise ValueError(f"No numeric token found in: {chunk!r}")
    return float(match.group(0))


def clean_name(name_field: str) -> str:
    name_raw = name_field.replace(r'^\mathrm{RV}', '^RV')
    name = re.sub(r'\\\^\{\s*\\mathrm\{RV\}\s*\}', '^RV', name_raw)
    name = re.sub(r'[{}\\$]', '', name).strip()
    return name


records = []
current = None
for idx, ln in enumerate(lines):
    parts = [p.strip() for p in ln.split('&')]
    if len(parts) < 11:
        print(f"Skipping line {idx} (expected 11 cols, got {len(parts)}): {ln}")
        continue

    name_field = parts[0]
    if name_field:  # value row
        try:
            name = clean_name(name_field)
            vals = [parse_number(p) for p in parts[1:11]]
            current = (name, vals)
        except ValueError as exc:
            print(f"Skipping value row {idx}: {exc}")
            current = None
    else:  # uncertainty row
        if current is None:
            print(f"Uncertainty row without preceding value row at line {idx}")
            continue
        try:
            errs = [parse_number(p) for p in parts[1:11]]
            name, vals = current
            records.append((name, vals, errs))
        except ValueError as exc:
            print(f"Skipping uncertainty row {idx}: {exc}")
        current = None

cols = [
    "P_d", "i_deg", "M1_Msun", "M2_Msun",
    "R1_Rsun", "R2_Rsun", "T1_K", "T2_K",
    "logg1", "logg2"
]
df = pd.DataFrame([
    {"name": name, **{c: v for c, v in zip(cols, vals)},
     **{f"{c}_err": e for c, e in zip(cols, errs)}}
    for name, vals, errs in records
])

print(f"Parsed {len(df)} systems")
df
````

````python
# Query SIMBAD per object using PTF names; keep bibcode table for reference
from astroquery.simbad import Simbad
from astropy.coordinates import SkyCoord
import astropy.units as u
import numpy as np

BIBCODE = "2018MNRAS.475.2560V"

simbad = Simbad()
simbad.TIMEOUT = 60
simbad.add_votable_fields('ra(d)', 'dec(d)', 'sp', 'ids', 'otypes')

# Reference: grab all objects associated with the bibcode (diagnostic only)
bib_objs = simbad.query_bibcode(BIBCODE)
print(f"Bibcode {BIBCODE}: {0 if bib_objs is None else len(bib_objs)} objects")
if bib_objs is not None:
    print(f"Bibcode columns: {bib_objs.colnames}")


def extract_fields(table_row):
    """Return (ra_deg, dec_deg, sp_type, ids_str, main_id)."""
    cols = table_row.colnames
    if 'RA_d' in cols and 'DEC_d' in cols:
        ra_deg = float(table_row['RA_d'])
        dec_deg = float(table_row['DEC_d'])
    else:
        ra_key = next((k for k in ('RA', 'ra', 'RAJ2000', 'RA_ICRS') if k in cols), None)
        dec_key = next((k for k in ('DEC', 'dec', 'DEJ2000', 'DECJ2000', 'DEC_ICRS') if k in cols), None)
        if ra_key and dec_key:
            coord = SkyCoord(table_row[ra_key], table_row[dec_key], unit=(u.hourangle, u.deg))
            ra_deg = coord.ra.deg
            dec_deg = coord.dec.deg
        else:
            raise KeyError(f"No RA/Dec columns found in row with cols {cols}")

    sp_val = None
    for key in ('SP_TYPE', 'sp_type', 'SP'):
        if key in cols:
            sp_val = table_row[key]
            break
    if sp_val is not None and hasattr(sp_val, 'mask') and getattr(sp_val, 'mask', False):
        sp_val = None
    sp_type = str(sp_val).strip() if sp_val not in (None, '') else None

    ids_str = str(table_row['IDS']) if 'IDS' in cols else ''
    main_id = str(table_row['MAIN_ID']) if 'MAIN_ID' in cols else None
    return ra_deg, dec_deg, sp_type, ids_str, main_id


ra_list, dec_list, sp_list, used_name, main_ids, ids_list = [], [], [], [], [], []

for idx, row in df.iterrows():
    candidates = []
    if pd.notna(row.get('simbad_name')):
        candidates.append(row['simbad_name'])
    if pd.notna(row.get('simbad_name_ptf')):
        candidates.append(row['simbad_name_ptf'])
    # Add aliases provided by user for better SIMBAD matching
    for base in candidates[:]:
        for alias in alias_map.get(base, []):
            candidates.append(alias)
    candidates.append(row['name'])

    # Deduplicate while preserving order
    seen = set()
    deduped = []
    for c in candidates:
        if c in seen:
            continue
        seen.add(c)
        deduped.append(c)
    candidates = deduped

    ra_deg = dec_deg = sp_type = ids_str = main_id = None
    matched = None

    for cand in candidates:
        try:
            result = simbad.query_object(cand)
        except Exception as e:
            print(f"{row['name']}: error querying '{cand}': {e}")
            continue

        if result is None or len(result) == 0:
            continue

        ra_deg, dec_deg, sp_type, ids_str, main_id = extract_fields(result[0])
        matched = cand
        break

    ra_list.append(ra_deg)
    dec_list.append(dec_deg)
    sp_list.append(sp_type)
    used_name.append(matched)
    main_ids.append(main_id)
    ids_list.append(ids_str)

    if matched is None:
        print(f"{row['name']}: not found in SIMBAD via {candidates}")
    else:
        msg = f"{row['name']} -> {matched}: RA={ra_deg:.6f}, Dec={dec_deg:.6f}"
        if sp_type:
            msg += f", SpType={sp_type}"
        print(msg)


# Overwrite columns with direct-query results
df['RA'] = ra_list
df['Dec'] = dec_list
df['SpType'] = sp_list
df['simbad_resolved_name'] = used_name
df['simbad_ids'] = ids_list
df['MAIN_ID'] = main_ids

print(f"\nCoords present for {np.sum(pd.notna(df['RA']))} systems")
print(f"Spectral types present for {np.sum(pd.notna(df['SpType']))} systems")
display(df[['name', 'simbad_resolved_name', 'RA', 'Dec', 'SpType']])
````
