````python
# ---- helper functions ----
columns = [
    "System Name", "RA", "Dec", "Period", "Eccentricity",
    "M1","M1_sin3i", "M2", "M2_sin3i", "q", "Mass Function",
    "Type1", "Type2", "Detection Method", "Reference", "Notes"
]
df = pd.DataFrame(columns=columns)

def add_observation(df, system_name,
                    ra, dec, period, ecc,
                    m1, m1_sin3i, m2, m2_sin3i, q, mass_func,
                    type1, type2, method, reference, notes=""):
    new_row = {
        "System Name": system_name,
        "RA": ra,
        "Dec": dec,
        "Period": period,
        "Eccentricity": ecc,
        "M1": m1,
        "M1_sin3i": m1_sin3i,
        "M2": m2,
        "M2_sin3i": m2_sin3i,
        "q": q,
        "Mass Function": mass_func,
        "Type1": type1,
        "Type2": type2,
        "Detection Method": method,
        "Reference": reference,
        "Notes": notes,
    }
    return pd.concat([df, pd.DataFrame([new_row])], ignore_index=True)

TRI_NAN = [np.nan, np.nan, np.nan]

# ---- Parsing utilities (robust to odd TeX) ----
def clean_cell(s: str) -> str:
    s = s.strip()
    # join broken "NGC \\ \n 2420-173" cases in pre-pass
    return s

def to_float_or_nan(s: str) -> float | float:
    t = s.strip()
    t = t.replace(',', '')  # just in case
    # remove TeX wrappers
    t = re.sub(r"[${}\\]", "", t)
    t = t.replace(r"\,", "")
    t = t.replace(r"\cdot", "")
    t = t.replace("–", "-").replace("—", "-")
    # blanks or dashes
    if t == "" or t == "-" or t == "–":
        return np.nan
    # sometimes "0." appears; that's fine
    try:
        return float(t)
    except Exception:
        # pick the first number found (handles things like '0.' or '7E-05')
        m = re.search(r"[-+]?\d+(?:\.\d+)?(?:[Ee][-+]?\d+)?", t)
        return float(m.group(0)) if m else np.nan

def parse_asym_mass_triplet(s: str):
    """
    Parse M_Ba field like:
      '1.9$+0.7\\atop-0.5$'  → [0.5, 1.9, 0.7]
      '{\\sl 1.3}$+0.3\\atop-0.2$' → [0.2, 1.3, 0.3]
      '2.3$+1.4\\atop-0.7$' → [0.7, 2.3, 1.4]
      '1.6' → [nan, 1.6, nan]
    """
    t = s.strip()
    # keep a copy for fallback
    # normalize: turn "\atop" into a separator, strip TeX, keep signs
    t = t.replace("\\,", "")
    t = t.replace("\\atop", "|")
    t = re.sub(r"{\\sl\s*", "", t)       # drop slanted markup start
    t = t.replace("{", "").replace("}", "")
    t = t.replace("$", "")
    t = t.replace("\\", "")
    # extract numbers (value, +err, -err)
    nums = re.findall(r"[-+]?\d+(?:\.\d+)?(?:[Ee][-+]?\d+)?", t)
    if len(nums) >= 3:
        val = float(nums[0])
        plus = abs(float(nums[1]))
        minus = abs(float(nums[2]))
        return [minus, val, plus]
    elif len(nums) >= 1:
        val = float(nums[0])
        return [np.nan, val, np.nan]
    else:
        return TRI_NAN.copy()

def as_triplet_val(x):
    """Value only → [nan, value, nan] triplet."""
    if x is None or (isinstance(x, float) and np.isnan(x)):
        return TRI_NAN.copy()
    try:
        v = float(x)
        return [np.nan, v, np.nan]
    except Exception:
        return TRI_NAN.copy()

def cleanup_name(cell0: str) -> str:
    """
    Turn the 'HD/DM' cell into a clean System Name:
     - ' 5424'  → 'HD 5424'
     - '$-64^\\circ$4333' → 'DM -64 4333'
     - 'NGC 2420-173' → as-is
    """
    raw = cell0.strip()
    # join DM pattern like $-64^\circ$4333
    m = re.match(r"\s*\$?\s*([+\-]?\d+)\s*\^\s*\\?circ\s*\$?\s*(\d+)", raw)
    if m:
        deg, num = m.group(1), m.group(2)
        return f"DM {deg} {num}"
    # strip TeX
    s = re.sub(r"[${}\\]", "", raw)
    s = s.replace("^circ", "")
    s = s.replace("^\circ", "")
    s = re.sub(r"\s+", " ", s).strip()
    if s.upper().startswith("NGC"):
        return s.replace(" \\\\", "").strip()
    # pure number → HD
    if re.fullmatch(r"\d+", s):
        return f"HD {s}"
    return s

def parse_ba_class(cell: str) -> str:
    # e.g., 'm', 's', 'm$\rightarrow$s'
    t = cell.strip()
    t = t.replace("$", "").replace("\\rightarrow", "->")
    return re.sub(r"\s+", "", t)
````

````python
# ---- Pull data rows from both parts of the table ----
# pre-join the split "NGC \\ \n 2420-173" pattern
latex_text = re.sub(r"NGC\s*\\\\\s*\n\s*", "NGC ", latex_text)

data_rows = []
for line in latex_text.splitlines():
    if "&" in line and "\\" in line and not line.strip().startswith("\\"):
        data_rows.append(line.strip())

# some data rows may wrap onto the next line (rare). Join if the last cell lacks '\\'
# (We already filtered lines with '\\', so we’re fine.)

# ---- Parse each row into fields we need ----
REF_BIBCODES = ['2019A&A...626A.127J']  # you asked to use this for all rows

for raw in data_rows:
    # split columns
    cols = [c.strip() for c in raw.split("&")]
    # make sure we have enough columns (table has 19)
    if len(cols) < 12:
        continue

    # clean trailing '\\'
    cols[-1] = cols[-1].split("\\")[0].strip()

    # Column indices in this table:
    # 0: HD/DM, 1: Ba class, 2: Teff, 3: log g, 4: Lmin, 5: L, 6: Lmax,
    # 7: M_Ba, 8: M_WD, 9: P(d), 10: e, 11: f(M),
    # 12: [Fe/H], 13: [Y/Fe], 14: [Zr/Fe], 15: [La/Fe], 16: [Ce/Fe], 17: [hs/ls], 18: Ref
    # Some rows may slightly shift around in earlier columns; but 7..12 are quite stable.

    try:
        name = cleanup_name(cols[0])
        ba_class = parse_ba_class(cols[1])

        M_Ba_trip = parse_asym_mass_triplet(cols[7])
        M_WD_val  = to_float_or_nan(cols[8])
        P_val     = to_float_or_nan(cols[9])
        e_val     = to_float_or_nan(cols[10])
        fM_val    = to_float_or_nan(cols[11])
        feh_val   = to_float_or_nan(cols[12]) if len(cols) > 12 else np.nan
        hsls_val  = to_float_or_nan(cols[17]) if len(cols) > 17 else np.nan

        # Build triplets 
        M1 = M_Ba_trip                      # M1 = Ba star (accretor)
        M2 = as_triplet_val(M_WD_val)       # M2 = WD (past donor)
        Period = as_triplet_val(P_val)
        Ecc    = as_triplet_val(e_val)
        fM     = as_triplet_val(fM_val)

        # Mass ratio q = M2/M1 = WD/Ba (central values only, no propagated errs)
        q_val = np.nan
        if not np.isnan(M_WD_val) and not np.isnan(M_Ba_trip[1]) and M_Ba_trip[1] != 0:
            q_val = M_WD_val / M_Ba_trip[1]
        q_trip = as_triplet_val(q_val)

        # Notes
        notes_bits = [f"Ba class: {ba_class}"]
        if not np.isnan(feh_val):  notes_bits.append(f"[Fe/H]={feh_val:.2f}")
        if not np.isnan(hsls_val): notes_bits.append(f"[hs/ls]={hsls_val:.2f}")
        notes_bits.append("Type mapping: M1=Ba (accretor), M2=WD (previous donor)")
        notes = "; ".join(notes_bits)


        # Add row (CORRECTED types)
        df = add_observation(
            df,
            system_name=name,
            ra=TRI_NAN.copy(), dec=TRI_NAN.copy(),
            period=Period, ecc=Ecc,
            m1=M1, m1_sin3i=TRI_NAN.copy(),
            m2=M2, m2_sin3i=TRI_NAN.copy(),
            q=q_trip, mass_func=fM,
            type1="barium star", type2="white dwarf",
            method=["RV"],
            reference=['2019A&A...626A.127J'],
            notes=notes
        )


    except Exception as ex:
        # Skip malformed line but print it so you can inspect if needed
        print(f"[WARN] Could not parse row:\n{raw}\nReason: {ex}\n")
````

````python
# ---- Save HDF5 (triplets as datasets, rest as JSON metadata) ----
def save_triplet_columns_and_metadata(df, triplet_columns, h5_filename):
    with h5py.File(h5_filename, "w") as f:
        for col in triplet_columns:
            try:
                # Coerce to numeric triplets
                arr = np.array(df[col].tolist(), dtype=float)
                if arr.ndim != 2 or arr.shape[1] != 3:
                    raise ValueError(f"{col} is not triplets")
                f.create_dataset(col, data=arr)
            except Exception as e:
                print(f"Skipping column '{col}' due to: {e}")
        meta_cols = [c for c in df.columns if c not in triplet_columns]
        meta_json = df[meta_cols].to_json(orient="records")
        f.create_dataset("metadata_json", data=np.bytes_(meta_json))


#  Run the function with metadata saving
triplet_cols = ["RA", "Dec", "Period", "Eccentricity", "M1", "M1_sin3i", "M2", "M2_sin3i", "q", "Mass Function",]

from paths import DATA_DIR

save_triplet_columns_and_metadata(df, triplet_cols, DATA_DIR / "result_tables/barium_stars_Jorissen2019.h5")

print(f"Built {len(df)} systems and saved to barium_stars.h5")
print(df[["System Name","Period","Eccentricity","M1","M2","q","Mass Function"]].head())
````
